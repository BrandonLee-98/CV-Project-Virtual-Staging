###01: Data Exploration & Environmental Analysis
Project: CV Virtual Staging

Objective: Analyze raw residential imagery to identify architectural challenges such as high-dynamic-range (HDR) lighting, transparency (windows), and structural empty-space geometry.

```python
import os
import PIL.Image as Image
import matplotlib.pyplot as plt

raw_dir = "../data/raw/"
images = [f for f in os.listdir(raw_dir) if f.endswith(('.jpg', '.png'))]

print(f"Total Raw Images: {len(images)}")
for img_name in images[:3]:
    with Image.open(os.path.join(raw_dir, img_name)) as img:
        print(f"File: {img_name} | Resolution: {img.size} | Mode: {img.mode}")
```

###Preprocessing Strategy
To prevent API timeouts and maintain aspect ratio integrity, all images are normalized to a maximum dimension of 1024px before being passed to the CV and Generative layers.
