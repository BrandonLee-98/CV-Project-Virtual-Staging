###The "Empty Room" Problem
Vacant rooms lack the furniture "anchors" standard YOLO models rely on. We implement YOLO-World (Zero-Shot) to identify architectural constants (Fireplaces, Fans, Windows) to provide context for the staging engine.

```python
# PROOF OF CONCEPT: Tuning thresholds for Architectural Anchors
def test_thresholds(image_path, conf_level):
    results = detector.predict(image_path, conf=conf_level)
    # Filter logic: Architecture is kept at lower conf than furniture
    anchors = ["window", "fireplace", "ceiling fan"]
    detections = [label for label, conf in zip(results.labels, results.confs)
                  if (label in anchors and conf > 0.10) or (conf > 0.25)]
    return detections

# Testing fireplace detection at 0.12 confidence
test_img = "../data/raw/placeholder_input3.jpg"
print(f"Detections at 0.12: {test_thresholds(test_img, 0.12)}")
```
